```python
import random

# ============================================================
# Part A - Machine Initialization
# ============================================================

machines = ['A', 'B', 'C']

M = 3
N_play = 10
epsilon = 0.2

# ============================================================
# Part B - Create Knowledge Tables
# ============================================================

# N = Number of times each machine is selected
N = {
    'A': 0,
    'B': 0,
    'C': 0
}

# R = Total reward received from each machine
R = {
    'A': 0,
    'B': 0,
    'C': 0
}

# Q = Estimated average reward of each machine
Q = {
    'A': 0.0,
    'B': 0.0,
    'C': 0.0
}

print("Initial Tables")
print("-" * 40)
print("N =", N)
print("R =", R)
print("Q =", Q)

# ============================================================
# Part C, D and E - Epsilon-Greedy + Reward + Q Update
# ============================================================

for play in range(1, N_play + 1):

    # Generate random number between 0 and 1
    r = random.random()

    # Find machine having highest Q-value
    highest_q = max(Q.values())

    # In case of a tie, choose the first machine
    highest_machine = max(machines, key=lambda x: Q[x])

    # --------------------------------------------------------
    # Modified Epsilon-Greedy Decision
    # --------------------------------------------------------

    if r < epsilon:
        # Exploration:
        # Choose randomly from machines except highest Q machine
        available_machines = [
            machine for machine in machines
            if machine != highest_machine
        ]

        selected_machine = random.choice(available_machines)

        decision = "Exploration"

    else:
        # Exploitation:
        # Choose machine with highest Q-value
        selected_machine = highest_machine

        decision = "Exploitation"

    # --------------------------------------------------------
    # Part D - Reward Collection
    # --------------------------------------------------------

    print("\nPlay", play)
    print("Random number =", round(r, 4))
    print("Decision =", decision)
    print("Play with machine", selected_machine)

    reward = float(input("Enter reward: "))

    # --------------------------------------------------------
    # Part E - Q-Value Update
    # --------------------------------------------------------

    # Increase participation count
    N[selected_machine] += 1

    # Add reward to total reward
    R[selected_machine] += reward

    # Incremental sample-average formula:
    #
    # Q(a) = Q(a) + (Reward - Q(a)) / N(a)
    #
    Q[selected_machine] = (
        Q[selected_machine]
        + (reward - Q[selected_machine]) / N[selected_machine]
    )

    print("Updated Q-value =", round(Q[selected_machine], 4))


# ============================================================
# Part F - Display Final Results
# ============================================================

print("\n")
print("=" * 65)
print("FINAL RESULTS")
print("=" * 65)

print(f"{'Machine':<12}{'N (Count)':<15}{'R (Total)':<15}{'Q-Value':<15}")
print("-" * 65)

for machine in machines:
    print(
        f"{machine:<12}"
        f"{N[machine]:<15}"
        f"{R[machine]:<15.2f}"
        f"{Q[machine]:<15.2f}"
    )

# Calculate total reward
total_reward = sum(R.values())

print("-" * 65)
print("Total Reward Received =", total_reward)

# Find best machine
best_machine = max(machines, key=lambda x: Q[x])

print("Best Machine =", best_machine)
print("Highest Q-Value =", round(Q[best_machine], 2))
```
